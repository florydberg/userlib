```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# PARAMETRI
# ----------------------------
csv_path = r"C:\Users\florydberg01\Contacts\Desktop\blue fluctuations power\Copy of power_461nm_19_12_25_box_transient.csv"

# ----------------------------
# LETTURA FILE
# ----------------------------
df = pd.read_csv(csv_path, sep=";")

time = df["[ms]"].values
power = df["[W]"].values


# ----------------------------
# PLOT
# ----------------------------
plt.figure(figsize=(9, 6))

plt.plot(time, power)
plt.xlabel("Tempo [ms]")
plt.ylabel("Potenza[W]")
plt.title("Fluttuazioni potenza")



plt.tight_layout()
plt.show()
```
