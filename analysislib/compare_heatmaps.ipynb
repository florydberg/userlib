```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Step 1: Load the two CSV files into DataFrames
df1 = pd.read_csv('2024-10-24_103533_ImagingFluo_Pow_FluoImaging_duration_tweezer_2.csv')
df2 = pd.read_csv('2024-10-24_103223_ImagingFluo_Pow_FluoImaging_duration_tweezer_1.csv')

# Step 2: Calculate the difference between the two DataFrames
# Make sure the DataFrames have the same shape
print(df1.shape)
print(df2.shape)
if df1.shape == df2.shape:
    diff_df = df1 - df2
else:
    raise ValueError("The two heatmaps must have the same dimensions")

# Step 3: Plot the heatmap of the differences
plt.figure(figsize=(10, 8))
sns.heatmap(diff_df, annot=True, cmap='coolwarm', center=0)

# Step 4: Display the heatmap
plt.title("Difference Between Two Heatmaps")
plt.show()
```
