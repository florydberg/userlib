```python

import h5py
import matplotlib.pyplot as plt
import os
import numpy as np


# List of folder paths containing HDF5 files
folder_paths = [
    "repumpers/0014",
    "repumpers/0015",
    "repumpers/0016",
    "repumpers/0017"
]
# Initialize lists to store data
loading_times = []  # Combined loading times for each folder
atom_numbers = []  # Combined atom number profiles for each folder
folder_names = []  # Store folder names for labeling

# Loop over the folders to extract and combine data
for folder_path in folder_paths:
    folder_name = os.path.basename(folder_path)  # Get the folder name for labeling
    combined_loading_time = []
    combined_atom_number = []
    
    # Get all HDF5 file paths in the folder
    file_paths = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith(".h5")]
    if not file_paths:
        print(f"No HDF5 files found in {folder_path}")
        continue
    
    for file_path in file_paths:
        with h5py.File(file_path, "r") as f:
            # Extract loading time and atom number datasets
            loading_time = f["loadTime_BlueMOT"][:]
            atom_number = f["atom_number"][:]
            
            combined_loading_time.append(loading_time)
            combined_atom_number.append(atom_number)
    
    # Combine data from all files in the folder (e.g., concatenate)
    combined_loading_time = np.concatenate(combined_loading_time)
    combined_atom_number = np.concatenate(combined_atom_number)
    
    loading_times.append(combined_loading_time)
    atom_numbers.append(combined_atom_number)
    folder_names.append(folder_name)

# Plotting the data
plt.figure(figsize=(10, 6))
for i in range(len(loading_times)):
    plt.plot(loading_times[i], atom_numbers[i], label=folder_names[i])

plt.xlabel("Loading Time")
plt.ylabel("Atom Number")
plt.title("Atom Number Profiles vs Loading Time")
plt.legend(title="Folders")
plt.grid(True)
plt.tight_layout()
plt.show()
```
